 **Data Claening**

18. Handling Missing Values

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Region"] = df["Region"].fillna("Unknown")
df["Shipping Status"] = df["Shipping Status"].fillna("Unknown")
```

20. Converting Order Date to Datetime Format

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
```

19. Verifying Missing Values After Cleaning

```python
df.isnull().sum().sum()
```

21. Checking Invalid Order Dates

```python
df["Order Date"].isnull().sum()
```

22. Saving the Cleaned Dataset

```python
df.to_csv("Cleaned_Ecommerce_Data.csv", index=False)
```

23. Importing Visualization Libraries

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

24. Visualizing Sales Distribution

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["Total Price"], bins=30, kde=True)
plt.title("Distribution of Total Sales")
plt.xlabel("Total Price")
plt.ylabel("Frequency")
plt.show()
```

25. Analyzing Sales by Region

```python
plt.figure(figsize=(8, 5))
df.groupby("Region")["Total Price"].sum().sort_values(ascending=False).plot(kind="bar")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

26. Checking Product Category Column Name

```python
df.columns.tolist()
```

27. Analyzing Sales by Product Category

```python
plt.figure(figsize=(10, 5))
df.groupby("Category")["Total Price"].sum().sort_values(ascending=False).plot(kind="bar")
plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

28. Analyzing Sales by Shipping Status

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Shipping Status", hue="Shipping Status", legend=False)
plt.title("Order Count by Shipping Status")
plt.xlabel("Shipping Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()
```

29. Analyzing Monthly Sales Trends

```python
df["Month"] = df["Order Date"].dt.to_period("M").astype(str)
monthly_sales = df.groupby("Month")["Total Price"].sum()

plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o")
plt.title("Monthly Sales Trends")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

30. Identifying Top 10 Products by Sales

```python
top_products = df.groupby("Product Name")["Total Price"].sum().nlargest(10)

plt.figure(figsize=(9, 7))
plt.pie(
    top_products,
    labels=top_products.index,
    autopct="%1.1f%%",
    startangle=140
)
plt.title("Top 10 Products by Sales Contribution")
plt.tight_layout()
plt.show()
```

31. Calculating Key Business Metrics

```python
print("Total Sales:", df["Total Price"].sum())
print("Average Sales:", df["Total Price"].mean())
print("Total Orders:", df.shape[0])
print("Total Customers:", df["Customer ID"].nunique())
print("Total Products:", df["Product Name"].nunique())
```

32. Relationship Between Quantity and Total Sales

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Quantity"], df["Total Price"], alpha=0.6)
plt.title("Quantity vs Total Sales")
plt.xlabel("Quantity")
plt.ylabel("Total Sales")
plt.show()
```
